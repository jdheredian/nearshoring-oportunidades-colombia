### Parámetros

La tabla se arma raspando dos páginas vivas: el listado de códigos ISO de iban.com y la tabla de
códigos de país de WITS del Banco Mundial. Ninguna de las dos versiona su contenido, por lo que
la salida de la descarga original, del 30 de julio de 2025, está archivada en
`data/raw/descargas/`.

La bandera `reejecutar_descarga` queda en falso para que la cadena use esa copia.

```python
reejecutar_descarga = False

url_iban = "https://www.iban.com/country-codes"
url_wits = "https://wits.worldbank.org/wits/wits/witshelp/content/codes/country_codes.htm"

encabezado_navegador = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"}
```

### Cruce de las dos nomenclaturas

Une ambas tablas por el código numérico de país, que es el identificador común.

```python
if reejecutar_descarga:
    import requests

    df_iso_iban = pd.read_html(url_iban)[0]
    df_iso_wits = pd.read_html(requests.get(url_wits, headers=encabezado_navegador).text, header=1)[0]

    df_codigos = df_iso_iban.merge(df_iso_wits, left_on="Numeric", right_on="Code", suffixes=("_iban", "_wits"))
else:
    df_codigos = pd.read_csv(raw / 'descargas' / 'country_codes.csv')

temp.mkdir(parents=True, exist_ok=True)
df_codigos.to_csv(temp / 'country_codes.csv', index=False)
print(f"países en la correspondencia: {len(df_codigos)}")
```
